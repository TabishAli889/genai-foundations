```python
from langchain.chat_models import init_chat_model   #LLM
from langchain.embeddings import init_embeddings    #Embeddings
# from langchain_community.document_loaders import PyPDFLoader
from langchain_core.documents import Document   #.txt
from langchain_text_splitters import RecursiveCharacterTextSplitter # chunks
from langchain_core.vectorstores import InMemoryVectorStore #vs
```

```python
text = open(file="../Documents/AI&ML-Lecture#3.txt", encoding="utf-8").read()
```

```python
type(text)
```

```python
doc = Document(page_content=text, metadata={"source": "../Documents/AI&ML-Lecture#3.txt"})
```

```python
doc
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=300, chunk_overlap=10)
```

```python
chunks = splitter.split_text(text=doc.page_content)
```

```python
len(chunks)
```

```python
embedd = init_embeddings(model="nomic-embed-text:latest", provider="ollama")
```

```python
vs = InMemoryVectorStore.from_documents(
    documents=chunks, 
    embedding=embedd,
    metadata=[{"source"}]
    )
```
